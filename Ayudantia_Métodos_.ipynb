```python
#Ejemplo 1
# Constante de IVA inicial
IVA = 0.12
precio_base = 14890 #este precio puede variar según lo que se quieraa calcular

# Calculamos precio con IVA
precio_total = precio_base * (1 + IVA)
print("Precio con IVA inicial:", precio_total)

# Cambiamos el IVA a 19%
IVA = 0.19
precio_total = precio_base * (1 + IVA)
print("Precio con IVA actualizado:", precio_total)
```

```python
#Ejemplo 1
#crear un lista
mascotas = ["gato", "perro", "pajaro", "pez"]
print(mascotas)

#Acceder a elementos de la lista
print(mascotas[0])
print(mascotas[3])#cabe recordar que en las listas el primer elemento es 0

#modificar elementos de la lista
mascotas[3] = "hamster"
print(mascotas)

#añadir elementos a la lista
mascotas.append("tortuga")
print(mascotas)

#eliminar elementos de la lista
mascotas.remove("gato")
print(mascotas)

#recorrer una lista
for mascota in mascotas:
    print(mascota)
```

```python
#Ejemplo 1
# Verificamos el rango de edad de una persona
edad = 25

if edad < 18:
    print("Eres menor de edad.")
elif edad >= 18 and edad <= 30:
    print("Eres joven adulto.")
elif edad > 30 and edad <= 60:
    print("Eres adulto.")
else:
    print("Eres adulto mayor.")
```

```python
#Ejemplo 2
# Verificamos si un número es positivo, negativo o par/impar
numero = -4

if numero > 0:
    if numero % 2 == 0:
        print("El número es positivo y par.")
    else:
        print("El número es positivo e impar.")
elif numero < 0:
    if numero % 2 == 0:
        print("El número es negativo y par.")
    else:
        print("El número es negativo e impar.")
else:
    print("El número es cero.")
```

```python
#Ejemplo 2
# Calcular la suma de los números ingresados hasta que el usuario ingrese un número negativo
suma = 0
numero = int(input("Ingresa un número (negativo para terminar): "))

while numero >= 0:
    suma += numero
    numero = int(input("Ingresa un número (negativo para terminar): "))

print("La suma total es:", suma)
```

6. **Ejemplo de aplicacion numpy**

```python
# Paso 1: Crear una matriz de tamaño 4x4 con números aleatorios entre 1 y 100
matriz = np.random.randint(1, 101, (4, 4))
print("Matriz original:")
print(matriz)

# Paso 2: Sumar todos los elementos de la matriz
suma_matriz = np.sum(matriz)
print(f"\nLa suma de todos los elementos de la matriz es: {suma_matriz}")

# Paso 3: Verificar si la suma es par o impar
if suma_matriz % 2 == 0:
    print("La suma es par.")

    # Paso 4: Crear una nueva matriz con los cuadrados de los números de la matriz original
    cuadrados = np.array([[x**2 for x in fila] for fila in matriz])
    print("\nMatriz con los cuadrados de los elementos:")
    print(cuadrados)

else:
    print("La suma es impar.")

    # Paso 5: Calcular el promedio de los elementos de la matriz
    promedio = np.mean(matriz)
    print(f"\nEl promedio de los elementos de la matriz es: {promedio:.2f}")
```
